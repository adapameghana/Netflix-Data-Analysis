```python
import pandas as pd

# Assuming you have already read the DataFrame as `df`
# Sample conversion for the "duration" column
def convert_duration(duration):
    if pd.isnull(duration):
        return None
    if 'min' in duration:
        return int(duration.split()[0])
    elif 'Season' in duration:
        # Assuming 1 season = 60 minutes as a placeholder (you can adjust as needed)
        return int(duration.split()[0]) * 60
    else:
        return None

# Apply the conversion function
df["duration in min"] = df["duration"].apply(convert_duration)
```

1 season = 60 minutes

```python
from collections import Counter
genres = df['listed_in'].dropna().str.split(', ')
flat_list = [genre for sublist in genres for genre in sublist]
genre_counts = Counter(flat_list).most_common(10)
genre_df = pd.DataFrame(genre_counts, columns=["Genre", "Count"])
plt.figure(figsize=(10, 6))
sns.barplot(data=genre_df, x="Count", y="Genre", palette='coolwarm')
plt.title("Top 10 Most Common Genres")
plt.show()
```

X-axis: Number of times each genre appears (Count).

Y-axis: Genre names.

The bars show the 10 most common genres in the dataset.

International Movies, Dramas, and Comedies are the top 3 most frequent genres.

The color gradient from the coolwarm palette provides a visual cue about the magnitude of the counts.

```python
import difflib

# Getting the movie name from the user
movie_name = input("Enter your favourite movie name:")

# Creating a list with all the movie names in the dataset
list_of_titles = movies_df["title"].tolist()

# Finding close matches
find_close_match = difflib.get_close_matches(movie_name, list_of_titles)

# Check if any match was found
if find_close_match:
    close_match = find_close_match[0]
    print("Best match found:", close_match)

    # Finding the index of the matched movie title
    matching_movies = movies_df[movies_df["title"] == close_match]
    index_of_the_movie = matching_movies.index[0]
    print("Index of the movie:", index_of_the_movie)
else:
    print(f"No close match found for '{movie_name}'")
```

# top 5 matches of the movies 

```python
import difflib

# Getting the movie name from the user
movie_name = input("Enter your favourite movie name: ")

# Creating a list with all the movie names in the dataset
list_of_titles = movies_df["title"].tolist()

# Finding top 5 close matches
top_matches = difflib.get_close_matches(movie_name, list_of_titles, n=5, cutoff=0.5)

# Check if any matches were found
if top_matches:
    print("\nTop 5 close matches found:")
    for i, match in enumerate(top_matches, start=1):
        print(f"{i}. {match}")

    # Optional: Get indices of the matched titles
    print("\nIndices of the matches:")
    for match in top_matches:
        index = movies_df[movies_df["title"] == match].index[0]
        print(f"{match} --> Index: {index}")
else:
    print("No close matches found for:", movie_name)
```

# top 5 movies with their details 

```python
import difflib

# Get user input
movie_name = input("Enter your favourite movie name: ")

# Create a list of movie titles
list_of_titles = movies_df["title"].tolist()

# Find top 5 best matches
top_matches = difflib.get_close_matches(movie_name, list_of_titles, n=5, cutoff=0.5)

# Check and display results
if top_matches:
    print("\nTop 5 close matches found:")
    for i, match in enumerate(top_matches, start=1):
        print(f"{i}. {match}")

    # Filter the DataFrame to show details of matched movies
    matched_movies_df = movies_df[movies_df["title"].isin(top_matches)]

    print("\nDetails of matched movies:\n")
    print(matched_movies_df)
else:
    print("No close matches found for:", movie_name)
```
